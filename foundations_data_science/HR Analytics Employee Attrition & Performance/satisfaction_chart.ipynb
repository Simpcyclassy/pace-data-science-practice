```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the datasets
employee_data = pd.read_csv("Employee.csv")
performance_data = pd.read_csv("PerformanceRating.csv")
satisfaction_data = pd.read_csv("SatisfiedLevel.csv")

# Step 1: Merge Employee.csv and PerformanceRating.csv on EmployeeID
employee_performance = pd.merge(employee_data, performance_data, on="EmployeeID")

# Step 2: Merge the result with SatisfiedLevel.csv on JobSatisfaction and SatisfactionID
merged_data = pd.merge(employee_performance, satisfaction_data, 
                       left_on="JobSatisfaction", right_on="SatisfactionID")

# Step 3: Plot Satisfaction Level vs. Attrition
plt.figure(figsize=(10, 6))
sns.boxplot(x=merged_data["Attrition"], y=merged_data["SatisfactionLevel"])
plt.title("Satisfaction Levels by Attrition Status", fontsize=16)
plt.xlabel("Attrition (Yes/No)", fontsize=12)
plt.ylabel("Satisfaction Level", fontsize=12)
plt.xticks(fontsize=10)
plt.yticks(fontsize=10)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```
