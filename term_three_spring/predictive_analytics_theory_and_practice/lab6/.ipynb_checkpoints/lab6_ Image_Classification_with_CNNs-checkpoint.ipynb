## **Step 1: Data Loading and Preprocessing**

In this section, I will:
- Download and explore the Cats vs Dogs dataset
- Set up data directories and organize the dataset
- Create data generators for training and validation
- Implement data augmentation techniques
- Visualize sample images to understand the data


```python
# Download and setup the Cats vs Dogs dataset
import urllib.request

# Set random seeds for reproducibility
tf.random.set_seed(42)
np.random.seed(42)
random.seed(42)

# Download the filtered cats vs dogs dataset
url = "https://storage.googleapis.com/mledu-datasets/cats_and_dogs_filtered.zip"
zip_path = "cats_and_dogs_filtered.zip"

if not os.path.exists(zip_path):
    print("Downloading cats vs dogs dataset...")
    urllib.request.urlretrieve(url, zip_path)
    print("Dataset downloaded successfully!")
else:
    print("Dataset already exists!")

# Extract the dataset
base_dir = "cats_and_dogs_filtered"
if not os.path.exists(base_dir):
    print("Extracting dataset...")
    with zipfile.ZipFile(zip_path, 'r') as zip_ref:
        zip_ref.extractall('.')
    print("Dataset extracted successfully!")
else:
    print("Dataset already extracted!")

# Define data directories
train_dir = os.path.join(base_dir, 'train')
validation_dir = os.path.join(base_dir, 'validation')

train_cats_dir = os.path.join(train_dir, 'cats')
train_dogs_dir = os.path.join(train_dir, 'dogs')
validation_cats_dir = os.path.join(validation_dir, 'cats')
validation_dogs_dir = os.path.join(validation_dir, 'dogs')

# Display dataset information
print(f"\nDataset Information:")
print(f"Training cats: {len(os.listdir(train_cats_dir))}")
print(f"Training dogs: {len(os.listdir(train_dogs_dir))}")
print(f"Validation cats: {len(os.listdir(validation_cats_dir))}")
print(f"Validation dogs: {len(os.listdir(validation_dogs_dir))}")
print(f"Total training images: {len(os.listdir(train_cats_dir)) + len(os.listdir(train_dogs_dir))}")
print(f"Total validation images: {len(os.listdir(validation_cats_dir)) + len(os.listdir(validation_dogs_dir))}")
```

```python
# Visualize sample images from the dataset
fig, axes = plt.subplots(2, 4, figsize=(12, 6))
fig.suptitle('Sample Images from Cats vs Dogs Dataset', fontsize=16, fontweight='bold')

# Sample cat images
cat_files = os.listdir(train_cats_dir)[:4]
for i, filename in enumerate(cat_files):
    img_path = os.path.join(train_cats_dir, filename)
    img = plt.imread(img_path)
    axes[0, i].imshow(img)
    axes[0, i].set_title(f'Cat {i+1}', fontweight='bold')
    axes[0, i].axis('off')

# Sample dog images  
dog_files = os.listdir(train_dogs_dir)[:4]
for i, filename in enumerate(dog_files):
    img_path = os.path.join(train_dogs_dir, filename)
    img = plt.imread(img_path)
    axes[1, i].imshow(img)
    axes[1, i].set_title(f'Dog {i+1}', fontweight='bold')
    axes[1, i].axis('off')

plt.tight_layout()
plt.show()

# Display image properties
sample_img_path = os.path.join(train_cats_dir, cat_files[0])
sample_img = plt.imread(sample_img_path)
print(f"\nSample Image Properties:")
print(f"Image shape: {sample_img.shape}")
print(f"Data type: {sample_img.dtype}")
print(f"Pixel value range: {sample_img.min()} - {sample_img.max()}")
```

```python
# Set up data generators with augmentation
# ImageDataGenerator already imported in dependencies cell

# Define image parameters
IMG_HEIGHT = 160
IMG_WIDTH = 160
BATCH_SIZE = 32
CLASS_NAMES = ['cats', 'dogs']

# Data augmentation for training data
train_datagen = ImageDataGenerator(
    rescale=1./255,           # Normalize pixel values to [0,1]
    rotation_range=20,        # Random rotation up to 20 degrees
    width_shift_range=0.1,    # Random horizontal shift
    height_shift_range=0.1,   # Random vertical shift
    shear_range=0.1,          # Random shear transformation
    zoom_range=0.1,           # Random zoom
    horizontal_flip=True,     # Random horizontal flip
    fill_mode='nearest'       # Fill mode for transformations
)

# Only rescaling for validation data (no augmentation)
validation_datagen = ImageDataGenerator(rescale=1./255)

# Create data generators with error handling
try:
    train_generator = train_datagen.flow_from_directory(
        train_dir,
        target_size=(IMG_HEIGHT, IMG_WIDTH),
        batch_size=BATCH_SIZE,
        class_mode='binary',      # Binary classification (cats vs dogs)
        seed=42
    )
    
    validation_generator = validation_datagen.flow_from_directory(
        validation_dir,
        target_size=(IMG_HEIGHT, IMG_WIDTH),
        batch_size=BATCH_SIZE,
        class_mode='binary',
        seed=42
    )
    
    print(f"\n✅ Data Generators Created Successfully:")
    print(f"Training samples: {train_generator.samples}")
    print(f"Validation samples: {validation_generator.samples}")
    print(f"Number of classes: {train_generator.num_classes}")
    print(f"Class indices: {train_generator.class_indices}")
    print(f"Image shape: ({IMG_HEIGHT}, {IMG_WIDTH}, 3)")
    print(f"Batch size: {BATCH_SIZE}")
    
except Exception as e:
    print(f"❌ Error creating data generators: {e}")
    print("Please check that the dataset directories exist and contain images.")
```

```python
# Visualize data augmentation effects
def show_augmented_images(generator, num_images=5):
    """Display original and augmented versions of images"""
    try:
        sample_batch, _ = next(generator)
        
        fig, axes = plt.subplots(1, num_images, figsize=(15, 3))
        fig.suptitle('Data Augmentation Examples', fontsize=16, fontweight='bold')
        
        for i in range(num_images):
            # Display augmented image
            axes[i].imshow(sample_batch[i])
            axes[i].set_title(f'Augmented Image {i+1}', fontweight='bold')
            axes[i].axis('off')
        
        plt.tight_layout()
        plt.show()
        return True
        
    except Exception as e:
        print(f"❌ Error displaying augmented images: {e}")
        return False

# Show augmented training images
print("🔄 Training Data Augmentation Examples:")
try:
    if 'train_generator' in locals():
        success = show_augmented_images(train_generator)
        if success:
            print("✅ Augmentation visualization completed successfully!")
    else:
        print("❌ Train generator not found. Please run the previous cell first.")
except Exception as e:
    print(f"❌ Error: {e}")

print("\n Step 1 Complete: Data Loading and Preprocessing")
print("Dataset loaded and preprocessed successfully!")
print("Images resized to 160x160 pixels")
print("Data augmentation applied to training set")
print("Data generators created for efficient batch processing")
```
