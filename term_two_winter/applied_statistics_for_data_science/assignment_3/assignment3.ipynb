## Question 1: Situation I - Exercise Habits Survey

**Context:**  
A survey was conducted on the exercise habits of college students. A random sample of 2,250 students was taken, comprising 1,180 males and 1,070 females. The study found that:  
- 825 male students exercise regularly.  
- 775 female students exercise regularly.  
- In total, 1,600 students exercise regularly.  
- For context, the average exercise time among these 1,600 students is 4 hours per week.

### 1.1 Point Estimation

#### 1.1.1 Proportion of Male Students Who Exercise Regularly 
  Pmale = Number of male exercisers/Total number of males = 825/1180 = 0.699
  
  Approximately **69.9%** of male college students exercise regularly.

#### 1.1.2 Proportion of Female Students Who Exercise Regularly
  Pfemale = Number of female exercisers/Total number of females = 775/1070 = 0.724
  
  Approximately **72.4% %** of female college students exercise regularly.

#### 1.1.3 Proportion of All College Students Who Exercise Regularly
   Poverall = Total exercisers/Total number of students = 1600/2250 = 0.711
   
  Approximately **71.11%** of all college students exercise regularly.

### 1.2 95% Confidence Interval for the Overall Proportion

To quantify the uncertainty around the overall point estimate, we construct a 95% confidence interval using the standard formula for proportions.

#### Step 1: Calculate the Standard Error (SE)
The standard error for a proportion where:
p = 0.711
n = 2250

SE = √(0.711×(1−0.711)=0.711×0.289 / 2250) ≈ 0.00956

#### Step 2: Determine the Margin of Error (ME)
For a 95% confidence interval, we use the z-value of 1.96:

ME=z×SE=1.96×0.00956≈0.01873

#### Step 3: Construct the Confidence Interval

- **Lower bound:**  0.711 - 0.01873 ≈ 0.6923
- **Upper bound:**  0.711 + 0.01873 ≈ 0.7297

**Interpretation:**  
We are 95% confident that the true proportion of college students who exercise regularly lies between approximately **69.2%** and **73.0%**.

## **Question 2: Situation 1 - Exercise Habits Survey for Community College Students**

**Context:**  
A study on exercise habits of college students was conducted in Manitoba. A random sample of 1,000 community college students was taken, and it was found that 215 received federal grants. Nationally, 25% of community college students are reported to receive federal grants. The Ministry of Education of Manitoba suspects that the proportion in Manitoba is lower.

### **Steps and Calculations**

1. **Hypotheses:**

   - **Null Hypothesis (H₀):** The proportion of Manitoba community college students receiving federal grants is 25% (p = 0.25).  
   - **Alternative Hypothesis (H₁):** The proportion is less than 25% (p < 0.25).

2. **Sample Proportion Calculation:**

   
   p= 215/1000 = 0.215
   

3. **Checking Conditions:**  
   - n x p_0 = 1000 x 0.25 = 250 >= 10  
   - n x (1-p_0) = 1000 x 0.75 = 750 >= 10  
   Conditions are met.

4. **Standard Error (SE):**

   SE = √(0.25x 0.75/1000) ≈ 0.01369

5. **Test Statistic (z-score):**

   z = (0.215 - 0.25)/0.01369 ≈ -2.555

6. **p-value:**  
   For a one-tailed test (p < 0.25), the p-value = P(Z < -2.555). From standard normal tables, this is approximately 0.0053.

7. **Conclusion:**  
   Since the p-value (≈0.0053) is less than the significance level α = 0.05, we **reject the null hypothesis**.  
   **Interpretation:** There is statistically significant evidence to conclude that the proportion of Manitoba community college students receiving federal grants is less than 25%.

## **Situation 3.1: Alcohol Consumption Among Liberal Arts Students**

**Context:**  
A recent social study reported that the average number of alcoholic drinks consumed per week by liberal arts college students is 12.5. However, the principal office of liberal arts college ABC believes that their students do not consume as much alcohol. A survey of 200 students from college ABC yielded:  
- Sample mean, (x) = 4.7 drinks per week  
- Sample standard deviation, s = 6.5 drinks

### **Step 1: Formulate the Hypotheses**

- **Null Hypothesis (H₀):** The true mean number of drinks consumed by the students is 12.5.  
  H₀: m = 12.5

- **Alternative Hypothesis (H₁):** The true mean is less than 12.5.  
  H₁: m < 12.5

### **Step 2: Check Conditions**

- The sample size is n = 200, which is large enough for the Central Limit Theorem to apply.  
- The data is assumed to be normally distributed or approximately normal.

### **Step 3: Calculate the Test Statistic**

1. **Compute the Standard Error (SE):**  
   SE = s/√n = 6.5/√200 ≈ 0.459

3. **Calculate the t-score:**  
   t = (x - m)/SE = (4.7 - 12.5)/0.459 ≈ -16.98

### **Step 4: Determine the p-value**

- For a one-tailed test with H₁: m < 12.5, the p-value is P(T < -16.98).  
- With such an extreme t-value, the p-value is effectively 0 (far below any common significance level such as 0.05).

### **Step 5: Conclusion**

- Since the p-value is nearly 0, we **reject the null hypothesis** at the 5% significance level.
- **Interpretation:** There is statistically significant evidence to conclude that the mean number of alcoholic drinks consumed per week by college ABC students is significantly less than 12.5.

## **Situation 3.2: Christmas Gift Expenditure by Canadian Families**

**Context:**  
A business study reported that, on average, Canadian families spend about $120 on Christmas gifts. However, the Department of Statistics Canada believes that families spend much more than $120. A survey of 2048 Canadian families found:  
- Sample mean, x = $225 
- Sample standard deviation, s = $80.45

### **Step 1: Formulate the Hypotheses**

- **Null Hypothesis (H₀):** The true mean expenditure is $120.  
  H₀: m = 120

- **Alternative Hypothesis (H₁):** The true mean expenditure is greater than \$120.  
  H₁: m > 120

### **Step 2: Check Conditions**

- The sample size is n = 2048, which is large enough for the Central Limit Theorem to apply.
- The data is assumed to be approximately normally distributed or the sample size is sufficient to rely on the CLT.

### **Step 3: Calculate the Test Statistic**

1. **Compute the Standard Error (SE):**  

   SE = s/√n = 80.45/√2048 ≈ 1.777

2. **Calculate the t-score:**  

   t = (x - m)/SE = (225 - 120)/1.777 ≈ 59.06

### **Step 4: Determine the p-value**

- For a one-tailed test with H₁: m > 120, the p-value is P(T > 59.06).
- With such an extremely high t-value, the p-value is effectively 0 (much lower than 0.05).

### **Step 5: Conclusion**

- Since the p-value is effectively 0, we **reject the null hypothesis** at the 5% significance level.
- **Interpretation:** There is overwhelming statistical evidence to conclude that Canadian families spend significantly more on Christmas gifts than $120.
