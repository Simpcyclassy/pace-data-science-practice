#### **Step 1: Impute Missing Values for `price` and `price_aprox_local_currency`**

### **7. Identifying and Handling Outliers**


### **Step 1: Calculate Outliers Using the IQR Method**

The IQR method is a commonly used technique to identify outliers. It works by calculating the difference between the 75th percentile (Q3) and the 25th percentile (Q1), then identifying values that lie outside the range of:
- **Q1 - 1.5 * IQR** (lower bound)
- **Q3 + 1.5 * IQR** (upper bound)

We'll apply this to numerical columns like `price`, `surface_total_in_m2`, and `rooms`, which are likely to have outliers.

#### **Code to Calculate Outliers Using IQR**

```python
# Calculate Q1 (25th percentile) and Q3 (75th percentile) for each of the columns
Q1_price = df['price'].quantile(0.25)
Q3_price = df['price'].quantile(0.75)
IQR_price = Q3_price - Q1_price

Q1_surface = df['surface_total_in_m2'].quantile(0.25)
Q3_surface = df['surface_total_in_m2'].quantile(0.75)
IQR_surface = Q3_surface - Q1_surface

Q1_rooms = df['rooms'].quantile(0.25)
Q3_rooms = df['rooms'].quantile(0.75)
IQR_rooms = Q3_rooms - Q1_rooms

# Calculate the lower and upper bounds for each column to identify outliers
lower_bound_price = Q1_price - 1.5 * IQR_price
upper_bound_price = Q3_price + 1.5 * IQR_price

lower_bound_surface = Q1_surface - 1.5 * IQR_surface
upper_bound_surface = Q3_surface + 1.5 * IQR_surface

lower_bound_rooms = Q1_rooms - 1.5 * IQR_rooms
upper_bound_rooms = Q3_rooms + 1.5 * IQR_rooms

# Identify outliers in the 'price', 'surface_total_in_m2', and 'rooms' columns
outliers_price = df[(df['price'] < lower_bound_price) | (df['price'] > upper_bound_price)]
outliers_surface = df[(df['surface_total_in_m2'] < lower_bound_surface) | (df['surface_total_in_m2'] > upper_bound_surface)]
outliers_rooms = df[(df['rooms'] < lower_bound_rooms) | (df['rooms'] > upper_bound_rooms)]

# Output the results
print(f"Number of outliers in price: {outliers_price.shape[0]}")
print(f"Number of outliers in surface area: {outliers_surface.shape[0]}")
print(f"Number of outliers in rooms: {outliers_rooms.shape[0]}")
```

### **Step 3: Handling Outliers Together**

Now that we've calculated and visualized the outliers, let's discuss how to handle them.

There are several approaches to dealing with outliers in this dataset. Since we’re not dropping any rows or columns, here are a few potential options for handling outliers:

1. **Capping Outliers**: 
   We could replace extreme outliers with a specified cap or floor value, such as the 1st and 99th percentiles. This way, we keep the majority of the data and eliminate the extreme values.

2. **Imputation of Outliers**: 
   Another approach is to replace outliers with a statistical value like the median or mean of the data.

3. **Transformations**:
   We could apply log transformations to columns like **price** and **surface area** to compress the range and reduce the impact of extreme values.


Based on the box plots and the number of outliers identified in the dataset, we have significant outliers in the **price**, **surface area**, and **rooms** columns. Here’s what the current analysis shows:

- **Price** has 454,412 outliers, which seems to represent a considerable portion of extreme values, suggesting that there might be a few properties that are extremely overpriced or underpriced compared to the majority.
- **Surface Area** has 1,298,547 outliers, indicating that there are potentially a large number of properties with unusual or extreme surface areas, likely much larger than the standard properties in the dataset.
- **Rooms** has 177,386 outliers, which indicates the presence of properties with an unusually high number of rooms.


### **Solution:**

- **Capping**: Set values that are higher than the 99th percentile to the 99th percentile and values lower than the 1st percentile to the 1st percentile.

Here’s the Python code to cap outliers for the three columns:

```python
# Capping outliers for 'price' column
lower_bound_price = df['price'].quantile(0.01)
upper_bound_price = df['price'].quantile(0.99)

df['price'] = df['price'].clip(lower=lower_bound_price, upper=upper_bound_price)

# Capping outliers for 'surface_total_in_m2' column
lower_bound_surface = df['surface_total_in_m2'].quantile(0.01)
upper_bound_surface = df['surface_total_in_m2'].quantile(0.99)

df['surface_total_in_m2'] = df['surface_total_in_m2'].clip(lower=lower_bound_surface, upper=upper_bound_surface)

# Capping outliers for 'rooms' column
lower_bound_rooms = df['rooms'].quantile(0.01)
upper_bound_rooms = df['rooms'].quantile(0.99)

df['rooms'] = df['rooms'].clip(lower=lower_bound_rooms, upper=upper_bound_rooms)
```

#### 5. **Price vs. Surface Area**

We can plot a scatter plot to examine the relationship between **price** and **surface area**.

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# Create bins for surface area (adjust the bin ranges as needed)
bins = [0, 50, 100, 150, 200, 250, 300, 350, 400, 500, df['surface_total_in_m2'].max()]
labels = [f"{bins[i]}-{bins[i+1]}" for i in range(len(bins)-1)]
df['surface_bin'] = pd.cut(df['surface_total_in_m2'], bins=bins, labels=labels, include_lowest=True)

# Group by surface area bin and calculate the average price
avg_price_surface = df.groupby('surface_bin')['price'].mean().reset_index()

plt.figure(figsize=(12, 6))
sns.barplot(x='surface_bin', y='price', data=avg_price_surface, palette='viridis', hue='price')
plt.title('Average Price vs. Surface Area Bins')
plt.xlabel('Surface Area (m²) (Binned)')
plt.ylabel('Average Price (BRL)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# --- Bar Chart for Price vs. Number of Rooms ---

```

#### **Price vs. Surface Area**

- **The density of smaller bins** (around 100 m²) aligns with the fact that most listings fall into a moderate size-and-price range, with only a limited number of properties occupying the largest size categories.
