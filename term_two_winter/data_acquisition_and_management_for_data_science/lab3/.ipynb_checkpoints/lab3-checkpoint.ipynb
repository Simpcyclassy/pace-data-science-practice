# **ETL with Bash**

## **Objectives**

The objective of this assignment is to implement an **ETL (Extract, Transform, Load) process using Bash scripting**. This exercise focuses on automating data extraction, transformation, and loading into a relational database using shell scripting techniques. The key learning goals are:

1. **Extracting Data**: Ingest and extract data from a remote source using Bash commands like `curl` or `wget`.
2. **Transforming Data**: Process and manipulate the extracted data using shell scripting tools (e.g., `cut`, `awk`, `sed`).
3. **Loading Data**: Import the transformed data into a **PostgreSQL database** using a Python script that creates a database table and loads the dataset.
4. **Scheduling the ETL Job**: Automate the ETL pipeline to run daily at a specific time using `cron` (job scheduler).

This assignment provides hands-on experience in **shell scripting, database interaction, data transformation, and job scheduling**, essential skills for data engineering roles.

## **Approach and Methodology**

The ETL pipeline is implemented using a structured approach that follows these key steps:

1. **Extract Phase:**
   - Use Bash commands (`curl` or `wget`) to download the dataset from a remote URL.
   - Implement error handling to ensure the dataset is successfully retrieved before proceeding.
   
2. **Transform Phase:**
   - Process the extracted dataset using tools like `cut`, `awk`, and `sed` to select specific columns and clean the data.
   - Ensure consistency and format the data appropriately for database insertion.
   
3. **Load Phase:**
   - Execute a Python script to connect to a PostgreSQL database and insert the transformed data.
   - The script will create the necessary table (if it does not exist) and load the cleaned dataset.
   
4. **Automation with Cron:**
   - Schedule the Bash script to run at a fixed time daily using `crontab`.
   - Ensure logs are generated for tracking execution success or failures.

This methodology ensures that the ETL process runs efficiently and can be easily maintained or modified for different datasets.

## **Task 1: Creating a Bash Script File and Setting Permissions**

### **Step 1: Create the Bash Script File**
To begin the ETL process, we create a Bash script file named `ETL_script.sh`:
```bash
touch ETL_script.sh
```
This command creates an empty file that will store the Bash commands for the ETL pipeline.

### **Step 2: Check File Permissions**
To verify the default permissions of the file, use the `ls -l` command:
```bash
ls -l ETL_script.sh
```
Output:
```
-rw-rw-r-- 1 simpcy simpcy 0 Feb 10 13:10 ETL_script.sh
```
Explanation:
- `rw-` → Owner has **read and write** permissions.
- `rw-` → Group members have **read and write** permission.
- `r--` → Others have **read-only** permission.
- The **execute (`x`) permission is missing**, meaning the script cannot be executed yet.

### **Step 3: Grant Execute Permission**
To allow the script to be executed, update its permissions using `chmod`:
```bash
chmod +x ETL_script.sh
```
This command adds **execute (`x`) permission** for the owner, making the script executable.

### **Step 4: Verify Updated Permissions**
Re-run the `ls -l` command to confirm the change:
```bash
ls -l ETL_script.sh
```
Expected output:
```
-rwxrwxr-x 1 simpcy simpcy 0 Feb 10 13:10 ETL_script.sh
```
Explanation:
- `rwx` → Owner can **read, write, and execute**.
- `rwx` → Group members can **read, write, and execute***.
- `r-x` → Others can **read and execute**.

## **Task 4: Automating Data Extraction in the Script**

### **Step 1: Implement Extraction in the Script**
To automate the extraction process, add the following lines inside `ETL_script.sh`:
To extract the dataset, use the `curl` or `wget` command to download the COVID-19 data from the provided URL:
```bash
# Extract Phase: Download dataset
DATA_URL="https://raw.githubusercontent.com/owid/covid-19-data/master/public/da>
RAW_DATA="raw_owid_covid_data.csv"
LOG_FILE="ETL_log.txt"

echo "$(date): Starting data extraction..." >> $LOG_FILE
curl -o $RAW_DATA $DATA_URL

    echo "$(date): Data extraction successful" >> $LOG_FILE
else
    echo "$(date): Data extraction failed" >> $LOG_FILE
fi
```

This command fetches the dataset and saves it as `raw_owid_covid_data.csv` in the current directory.

## **Task 5: Transforming Data**

### **Step 1: Extract Specific Columns**
The transformation step involves selecting specific columns (1, 2, 4, and 5) from the dataset. This can be achieved using the `cut` command:
```bash
cut -d',' -f1,2,4,5 raw_owid-covid-data.csv > transformed_data.csv
```
- The `-d','` option specifies that the **delimiter** is a comma.
- The `-f1,2,4,5` option selects only **columns 1, 2, 4, and 5**.
- The output is saved to `transformed_data.csv`.

### **Step 2: Save the Transformation Process in the Script**
Modify `ETL_script.sh` and add the transformation step inside it:
```bash
# Transform Phase: Select specific columns
TRANSFORMED_DATA="transformed_data.csv"
echo "$(date): Starting data transformation..." >> $LOG_FILE
cut -d',' -f1,2,4,5 $RAW_DATA > $TRANSFORMED_DATA

# Verify if transformation was successful
if [[ -f "$TRANSFORMED_DATA" ]]; then
    echo "$(date): Data transformation successful" >> $LOG_FILE
else
    echo "$(date): Data transformation failed" >> $LOG_FILE
fi
```

## **Task 6: Loading Data into PostgreSQL**

### **Step 1: Create a Python Script for Data Loading**
To load the transformed data into a PostgreSQL database, we use a Python script that:
1. Connects to a PostgreSQL instance.
2. Creates a database (`covid_db`) and a table (`COVID`) if they do not exist.
3. Loads the transformed data using the `COPY` command.

Created a new Python script named `load_to_postgres.py` and add the following code:

```python
import psycopg2
import os
import sys

# Define database connection parameters
db_name = os.getenv('DB_NAME', 'covid_db')
db_user = os.getenv('DB_USER', 'username')
db_password = os.getenv('DB_PASSWORD', 'password')
db_host = os.getenv('DB_HOST', 'localhost')
db_port = os.getenv('DB_PORT', '5432')

# Define input file
transformed_data = "transformed_data.csv"

conn = None
cursor = None

try:
    # Connect to PostgreSQL default database to check/create covid_db
    temp_conn = psycopg2.connect(
        dbname="postgres", user=db_user, password=db_password, host=db_host, port=db_port
    )
    temp_conn.autocommit = True
    temp_cursor = temp_conn.cursor()

    # Create the database if it doesn't exist
    temp_cursor.execute(f"SELECT 1 FROM pg_database WHERE datname = '{db_name}'")
    if not temp_cursor.fetchone():
        temp_cursor.execute(f"CREATE DATABASE {db_name}")

    temp_cursor.close()
    temp_conn.close()

    # Now, connect to the actual database
    conn = psycopg2.connect(
        dbname=db_name, user=db_user, password=db_password, host=db_host, port=db_port
    )
    cursor = conn.cursor()

    # Create table if it doesn't exist
    cursor.execute("""
        CREATE TABLE IF NOT EXISTS COVID (
            iso_code VARCHAR(10),
            continent VARCHAR(50),
            date DATE,
            total_cases BIGINT
        );
    """)

    # Load data into the table
    with open(transformed_data, 'r') as file:
        cursor.copy_expert("COPY COVID FROM STDIN WITH CSV HEADER", file)
    
    conn.commit()
    print("Data successfully loaded into PostgreSQL.")

except psycopg2.OperationalError as e:
    print("Database connection error:", e)
    sys.exit(1)
except Exception as e:
    print("Error loading data:", e)
    sys.exit(1)
finally:
    if cursor:
        cursor.close()
    if conn:
        conn.close()
```

### **Step 2: Modify ETL Script to Call Python Loader**
Update `ETL_script.sh` to include the database loading step:

```bash
# Load Phase: Load transformed data into PostgreSQL
echo "$(date): Starting data load into PostgreSQL..." >> $LOG_FILE
python3 load_to_postgres.py

if [[ $? -eq 0 ]]; then
    echo "$(date): Data successfully loaded into PostgreSQL" >> $LOG_FILE
else
    echo "$(date): Data load failed" >> $LOG_FILE
fi
```

## **Task 8: Automating the ETL Process with Cron**

### **Step 1: Open the Crontab Editor**
To schedule the ETL script to run daily at **10 minutes past midnight (00:10 AM)**, I opened the crontab editor by running:
```bash
crontab -e
```

### **Step 2: Add the Cron Job**
Append the following line to schedule the script execution:
```bash
10 0 * * * /path/to/ETL_script.sh >> /path/to/etl_log.txt 2>&1
```
#### **Explanation:**
- `10 0 * * *` → Runs the script at **12:10 AM** every day.
- `/path/to/ETL_script.sh` → path to my ETL script.
- `>> /path/to/etl_log.txt 2>&1` → Redirects both **output and errors** to the log file for debugging.

### **Step 3: Verify the Cron Job**
To confirm that the cron job has been added:
```bash
crontab -l
```
This should display the scheduled job.
