## **✅ Step 1: Implement `create_directories.py` (Utility File)**
We will write a separate utility script to manage directory creation.  
📍 **Path:** `~/airflow/dags/utils/create_directories.py`  

```python
import os
import logging
from airflow.utils.log.logging_mixin import LoggingMixin

# Use Airflow's logging system
log = LoggingMixin().log

def create_directories():
    """Dynamically creates required directories for the ETL pipeline and logs the process."""

    # Retrieve base path from environment variable or use the default
    base_path = os.getenv(
        'BASE_PATH',
        "/Users/chiomaonyekpere/Documents/PACE/data-science-python-practice/term_two_winter/data_acquisition_and_management_for_data_science/lab4"
    )

    # Define required directories using environment variables (fallback to defaults)
    directories = {
        'RAW_DATA_DIR': os.getenv('RAW_DATA_DIR', os.path.join(base_path, 'raw_data')),
        'PROCESSED_DATA_DIR': os.getenv('PROCESSED_DATA_DIR', os.path.join(base_path, 'processed_data')),
        'LOG_DIR': os.getenv('LOG_DIR', os.path.join(base_path, 'logs'))
    }

    # Ensure log directory exists
    os.makedirs(directories['LOG_DIR'], exist_ok=True)

    # Define log file path
    log_file_path = os.path.join(directories['LOG_DIR'], 'setup.log')

    # Configure logging to write to the log file (Explicit FileHandler approach)
    file_handler = logging.FileHandler(log_file_path)
    file_handler.setLevel(logging.INFO)
    file_formatter = logging.Formatter('%(asctime)s [%(levelname)s] %(message)s')
    file_handler.setFormatter(file_formatter)

    # Add the file handler to the logger
    log.addHandler(file_handler)

    # Also log to Airflow task logs
    log.info(f"Logging to: {log_file_path}")

    # Create directories and log messages
    for name, path in directories.items():
        try:
            os.makedirs(path, exist_ok=True)
            if os.path.exists(path):
                log.info(f"✅ Successfully created: {path} ({name})")
            else:
                log.error(f"❌ Failed to create: {path}")
        except Exception as e:
            log.error(f"❌ Error creating {path}: {str(e)}")

    log.info("✅ Directory structure set up successfully!")
```

### **Python Script for Downloading Data**

#### **`download_data.py`**
```python
import os
import requests
import logging
from airflow.utils.log.logging_mixin import LoggingMixin

# Use Airflow's logging system
log = LoggingMixin().log

def download_traffic_data():
    """Downloads traffic data from a remote URL and saves it to the raw_data directory."""

    # Retrieve paths from environment variables or defaults
    base_path = os.getenv(
        'BASE_PATH',
        "/Users/chiomaonyekpere/Documents/PACE/data-science-python-practice/term_two_winter/data_acquisition_and_management_for_data_science/lab4"
    )
    raw_data_dir = os.getenv('RAW_DATA_DIR', os.path.join(base_path, 'raw_data'))
    
    # Ensure raw_data directory exists
    os.makedirs(raw_data_dir, exist_ok=True)

    # Define dataset URL and destination file path
    data_url = "https://example.com/traffic_data.csv"  # Replace with actual URL
    output_file = os.path.join(raw_data_dir, "traffic_data.csv")

    log.info(f"Starting download from {data_url}...")
    
    try:
        response = requests.get(data_url, stream=True, timeout=30)
        response.raise_for_status()  # Raise an error for bad responses (4xx, 5xx)

        # Write the file
        with open(output_file, "wb") as file:
            for chunk in response.iter_content(chunk_size=1024):
                file.write(chunk)

        log.info(f"✅ Download successful: {output_file}")
    except requests.exceptions.RequestException as e:
        log.error(f"❌ Download failed: {str(e)}")
        raise

```

## **4. Task 4: Extracting Data - Downloading the File**

### **Step 1: Define the Download Process**
The Extract phase involves downloading the latest version of the dataset each time the script is executed. This ensures that the data used in the pipeline is always up to date. The dataset is downloaded from the specified remote location.

### **Step 2: Implement the Download Script**
The following script performs the extraction step:

```bash
# Extract Phase: Download dataset
DATA_URL="https://raw.githubusercontent.com/owid/covid-19-data/master/public/data/owid-covid-data.csv"
RAW_DATA="raw_owid_covid_data.csv"
TRANSFORMED_DATA="transformed_covid_data.csv"
LOG_FILE="ETL_log.txt"

# Log the start of the extraction process
echo "$(date): Starting data extraction..." >> $LOG_FILE

# Download the dataset (overwrite existing file to get the latest data)
curl -o $RAW_DATA $DATA_URL

# Verify if the file was downloaded successfully
if [[ -f "$RAW_DATA" ]]; then
    ROW_COUNT=$(($(wc -l < "$RAW_DATA") - 1))
COL_COUNT=$(head -1 "$RAW_DATA" | awk -F',' '{print NF}')
echo "$(date): Data extraction successful. Rows extracted: $ROW_COUNT, Columns extracted: $COL_COUNT" >> $LOG_FILE
else
    echo "$(date): Data extraction failed" >> $LOG_FILE
    exit 1
fi
```

### **Explanation of the Script**
1. **Defines Variables**:
   - `DATA_URL` stores the dataset's URL.
   - `RAW_DATA` defines the filename for storing the extracted data.
   - `LOG_FILE` keeps track of execution steps.
2. **Logs the Start of Extraction**: The script logs when the download starts.
3. **Downloads the File**: The `curl` command fetches the dataset and overwrites any previous file.
4. **Verifies the Download**:
   - If the file exists after the download, it logs success.
   - If the file is missing, it logs the failure and exits the script to prevent further execution errors.

## **Task 6: Loading Data into PostgreSQL**

### **Step 1: Create a Python Script for Data Loading**
To load the transformed data into a PostgreSQL database, we use a Python script that:
1. Connects to a PostgreSQL instance.
2. Creates a database (`covid_db`) and a table (`COVID`) if they do not exist.
3. Loads the transformed data using the `COPY` command.

Created a new Python script named `load_to_postgres.py` and add the following code:

```python
import psycopg2
import os
import sys

# Define database connection parameters
db_name = os.getenv('DB_NAME', 'covid_db')
db_user = os.getenv('DB_USER', 'username')
db_password = os.getenv('DB_PASSWORD', 'password')
db_host = os.getenv('DB_HOST', 'localhost')
db_port = os.getenv('DB_PORT', '5432')

# Define input file
transformed_data = "transformed_data.csv"

conn = None
cursor = None

try:
    # Connect to PostgreSQL default database to check/create covid_db
    temp_conn = psycopg2.connect(
        dbname="postgres", user=db_user, password=db_password, host=db_host, port=db_port
    )
    temp_conn.autocommit = True
    temp_cursor = temp_conn.cursor()

    # Create the database if it doesn't exist
    temp_cursor.execute(f"SELECT 1 FROM pg_database WHERE datname = '{db_name}'")
    if not temp_cursor.fetchone():
        temp_cursor.execute(f"CREATE DATABASE {db_name}")

    temp_cursor.close()
    temp_conn.close()

    # Now, connect to the actual database
    conn = psycopg2.connect(
        dbname=db_name, user=db_user, password=db_password, host=db_host, port=db_port
    )
    cursor = conn.cursor()

    # Create table if it doesn't exist
    cursor.execute("""
        CREATE TABLE IF NOT EXISTS COVID (
            iso_code VARCHAR(10),
            continent VARCHAR(50),
            date DATE,
            total_cases BIGINT
        );
    """)

    # Load data into the table
    with open(transformed_data, 'r') as file:
        cursor.copy_expert("COPY COVID FROM STDIN WITH CSV HEADER", file)
    
    conn.commit()
    print("Data successfully loaded into PostgreSQL.")

except psycopg2.OperationalError as e:
    print("Database connection error:", e)
    sys.exit(1)
except Exception as e:
    print("Error loading data:", e)
    sys.exit(1)
finally:
    if cursor:
        cursor.close()
    if conn:
        conn.close()
```

## **6. Task 6: Loading Data into PostgreSQL**

### **Step 1: Create a Python Script for Data Loading**
To load the transformed data into a PostgreSQL database, we use a Python script that:
1. Connects to a PostgreSQL instance.
2. Creates a database (`covid_db`) and a table (`COVID`) if they do not exist.
3. Loads the transformed data using the `COPY` command.

Create a new Python script named `load_to_postgres.py` and add the following code:

```python
import psycopg2
import os
import sys
from datetime import datetime

# Define database connection parameters
db_name = os.getenv('DB_NAME', 'covid_db')
db_user = os.getenv('PSQL_USER', 'postgres')
db_password = os.getenv('DB_PASSWORD', 'postgres')
db_host = os.getenv('DB_HOST', 'localhost')
db_port = os.getenv('DB_PORT', '5432')

# Define input file
transformed_data = "transformed_covid_data.csv"

conn = None
cursor = None

try:
    # Connect to PostgreSQL default database to check/create covid_db
    temp_conn = psycopg2.connect(
        dbname="postgres", user=db_user, password=db_password, host=db_host, port=db_port
    )
    temp_conn.autocommit = True
    temp_cursor = temp_conn.cursor()

    # Create the database if it doesn't exist
    temp_cursor.execute(f"SELECT 1 FROM pg_database WHERE datname = '{db_name}'")
    if not temp_cursor.fetchone():
        temp_cursor.execute(f"CREATE DATABASE {db_name}")

        cursor.copy_expert("COPY COVID FROM STDIN WITH CSV HEADER", file)

    # Check row count after insertion
    cursor.execute("SELECT COUNT(*) FROM COVID;")
    after_insert = cursor.fetchone()[0]
    rows_inserted = after_insert

    conn.commit()
    print(f"Data successfully loaded. {rows_inserted} rows inserted into {db_name}. Database now has {after_insert} rows.")

    # Log the inserted row count

    with open("ETL_log.txt", "a") as log_file:
        log_file.write(f"{datetime.now()}: Data successfully loaded. {rows_inserted} rows inserted. Database now has {after_insert} rows.\n")
    
    conn.commit()
    print("Data successfully loaded into PostgreSQL.")

except psycopg2.OperationalError as e:
    print("Database connection error:", e)
    sys.exit(1)
except Exception as e:
    print("Error loading data:", e)
    sys.exit(1)
finally:
    if cursor:
        cursor.close()
    if conn:
        conn.close()
```

Append the following line to schedule the script execution:
```bash

```

- `10 0 * * *` → Runs the script at **12:10 AM** every day.
- `/path/to/ETL_script.sh` → the **absolute path** to my ETL script.
- `>> /path/to/etl_log.txt 2>&1` → Redirects both **output and errors** to the log file for debugging.

## **Conclusion**

This assignment successfully demonstrated the implementation of an **ETL (Extract, Transform, Load) pipeline** using **Bash scripting and PostgreSQL**. Throughout the process, we automated the extraction of **real-world COVID-19 data**, transformed it by selecting relevant fields, and efficiently loaded it into a structured database for analysis.

### **Key Achievements**
1. **Robust Data Extraction**  
   - The pipeline ensures that the latest dataset is always downloaded, preventing outdated data from being processed.
   - The number of **extracted rows and columns** is logged, improving transparency.

2. **Efficient Data Transformation**  
   - The script processes only relevant columns (`iso_code`, `continent`, `date`, `total_cases`), optimizing storage and query performance.
   - The transformed dataset is validated before loading, ensuring consistency.
   - The transformation step now logs both **rows transformed** and **columns retained**, improving monitoring.

3. **Reliable Data Loading into PostgreSQL**  
   - The database creation is **idempotent** (it checks if the database and table exist before creating them).
   - The **database is cleared before reloading data**, preventing duplication issues.
   - The script logs the **number of rows inserted** and the **total row count after insertion**, allowing for better tracking.

4. **Comprehensive Logging for Debugging and Monitoring**  
   - The ETL process logs every key step, including:
     - **Download success/failure**
     - **Number of extracted and transformed rows**
     - **Success/failure of data load**
     - **Row count before and after insertion in the database**
   - This ensures transparency and easier debugging.

5. **Automation with Cron Jobs**  
   - The ETL process is fully **automated using a cron job**, ensuring that the pipeline runs **daily at 12:10 AM**.
   - The job is logged for verification, and system logs are monitored for execution tracking.
   - Environment variables were explicitly set to ensure the cron job runs as expected.

### **Challenges and Fixes**
- **Initial Cron Job Issues** → Debugging logs were enabled, and execution paths were explicitly defined.
- **Data Overwriting Concerns** → Implemented **idempotent data loading** to clear old data before new data insertion.
- **Logging Enhancements** → Now logs **row counts, error messages, and system statuses**, providing better monitoring.

### **Future Enhancements**
To further improve the ETL pipeline, we can:
- Implement **data validation** before inserting records into the database.
- Add **error notifications** (e.g., email alerts) when failures occur.
- Optimize performance by using **batch inserts** instead of row-by-row insertion.
- Extend the pipeline for **incremental updates** rather than full database reloads.

### **Final Thoughts**
This ETL workflow provides a **scalable and automated solution** for handling data ingestion, transformation, and storage. By leveraging **Bash scripting, Python, and PostgreSQL**, we have created a **structured, repeatable, and efficient data pipeline**.
