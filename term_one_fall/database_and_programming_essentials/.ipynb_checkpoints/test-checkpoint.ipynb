```python
import os
import mysql.connector as sql
os.environ['MYSQL_PASSWORD'] = "FuckAdam2!"
password = os.getenv('MYSQL_PASSWORD')
mydb = sql.connect(
    host = "localhost",
    user = "root",
    password = f"{password}",
    use_pure = "true"
)
print(f"Mysql is connected successfully: {mydb}")
```

```python
# create cursor object

mycursor = mydb.cursor()
mycursor.execute("DROP DATABASE IF EXISTS testDB")
mycursor.execute("CREATE DATABASE testDB")
print("Database testDB has been dropped and created.")
```

```python
mycursor.execute("SHOW DATABASES")
for db in mycursor:
    print(db)
```

```python
mycursor.execute("USE testDB")
mycursor.execute("CREATE TABLE student(id INT PRIMARY KEY, name VARCHAR(50) NOT NULL, age INT);")
```

```python
mycursor.execute("SHOW TABLES;")
for tb in mycursor:
    print(tb)
```

```python
sql_insert = "INSERT INTO student (id, name, age) VALUES(%s, %s, %s);"
student1 = (1011, 'John', 26)
mycursor.execute(sql_insert, student1)
mydb.commit()
```

```python
sql_insert = "INSERT INTO student (id, name, age) VALUES(%s, %s, %s);"
student_data = [(1012, "Bob", 26), (1013, "Mary", 32), (1014, "John", 29)]
mycursor.executemany(sql_insert, student_data)
mydb.commit()
```

```python
mycursor.execute("SELECT * FROM student;")
myresults = mycursor.fetchall()
for row in myresults:
    print(row)
mydb.commit()
```

```python
mycursor.execute("SELECT name, age FROM student WHERE age > 30;")
myresults = mycursor.fetchall()
for row in myresults :
    print(row)
mydb.commit()
```

```python
from tabulate import tabulate
print(tabulate(myresults, headers=['Name', 'Age'], tablefmt='psql'))
```
