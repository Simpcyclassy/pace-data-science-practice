```python
import os
import csv
import mysql.connector as sql

os.environ['MYSQL_PASSWORD'] = "12345678"
password = os.getenv('MYSQL_PASSWORD')

mydb = sql.connect(
    host = "localhost",
    user = "root",
    password = f"{password}",
    database= "testDB",
    use_pure = "true"
)
print(f"Mysql is connected successfully: {mydb}")
```

```python
mycursor = mydb.cursor()
mycursor.execute("DROP TABLE IF EXISTS student_testcsv;")
mycursor.execute("CREATE TABLE student_testcsv(ID INT PRIMARY KEY, Name VARCHAR(50) NOT NULL, Age INT);")
```

```python
csv_data = csv.reader(open('sample_student_data.csv'))
next(csv_data)
for row in csv_data:
    mycursor.execute("INSERT INTO student_testcsv(ID, Name, Age) VALUES(%s, %s, %s)", row)
mydb.commit()
```

```python
mycursor.execute("SELECT * FROM student_testcsv;")
myresults = mycursor.fetchall()
for row in myresults:
    print(row)
mydb.commit()
```
