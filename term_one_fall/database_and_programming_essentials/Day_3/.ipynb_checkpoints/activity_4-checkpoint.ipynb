```python
import pandas as pd
df= pd.read_csv('orders.csv',na_values =['Not Available', 'Unknown'])
df.head(30)
df['Customer Status'].unique()
```

```python
df.rename(columns={'Customer ID':'customer_id', 'Customer Status':'customer_status'})
```

```python
df.columns = df.columns.str.replace(' ','_')
df.head(10)
```

```python
df.columns = df.columns.str.lower()
df.head(10)
```

```python
df['total_cost_price'] = df['cost_price_per_unit'] * df['quantity_ordered']
df.head(10)
```

```python
df['profit'] = df['total_retail_price_for_this_order'] - df['total_cost_price']
df.head(10)
```

```python
df['order_date'] = pd.to_datetime(df['date_order_was_placed'],format="%d-%b-%y")
df.head(10)
```

```python
df.drop(columns=['cost_price_per_unit', 'date_order_was_placed'], inplace = True)
df.head(10)
```

```python
duplicateRows = df[df.duplicated()]
df.drop_duplicates(keep=False)
df.head(10)
```

```python
from sqlalchemy import create_engine
import os

os.environ['MYSQL_PASSWORD'] = "12345678"
password = os.getenv('MYSQL_PASSWORD')

# Create engine and handle connection errors
engine = create_engine(f'mysql+mysqlconnector://root:{password}@localhost/testDB',
            connect_args={'auth_plugin': 'caching_sha2_password'})
print(f"Mysql is connected successfully with create_engine: {engine}")
df.to_sql('orders', engine, if_exists ='replace', index=False)
```

```python
# Find top 10 highest sales along with the customer IDs.

query = """
    SELECT customer_id, sum(total_retail_price_for_this_order) AS sales
    FROM orders
    GROUP BY customer_id
    ORDER BY sales DESC
    LIMIT 10;
"""

df = pd.read_sql(query, engine)
df.head(30)
```

```python
# Find month over month growth comparison for 2020 and 2021 sales, e.g., Jan. 2020 vs Jan. 2021
query = """
    WITH CTE1 AS
        (SELECT DISTINCT year(order_date) AS order_year, month(order_date) AS
            order_month, sum(total_retail_price_for_this_order) AS sales
            FROM orders
            GROUP BY year(order_date), month(order_date)
            ORDER BY year(order_date), month(order_date)
        )
    SELECT CTE1.order_month,
        sum(CASE WHEN CTE1.order_year = 2020 THEN sales ELSE 0 END) AS sales_2020,
        sum(CASE WHEN CTE1.order_year = 2021 THEN sales ELSE 0 END) AS sales_2021
        FROM CTE1
        GROUP BY CTE1.order_month
        ORDER BY CTE1.order_month;
"""

df = pd.read_sql(query, engine)

df.head(30)
```
