```python
import xml.etree.ElementTree as ET
import os
import mysql.connector as sql

# Parse XML file
tree = ET.parse('employee.xml')
root = tree.getroot()

os.environ['MYSQL_PASSWORD'] = "12345678"
password = os.getenv('MYSQL_PASSWORD')

mydb = sql.connect(
    host = "localhost",
    user = "root",
    password = f"{password}",
    use_pure = "true"
)
print(f"Mysql is connected successfully: {mydb}")

mycursor = mydb.cursor()
mycursor.execute("use day5;")
```

```python
for employee in root.findall('employee'):
    employee_name = employee.find('employee_name').text
    department = employee.find('department').text
    salary = employee.find('salary').text
    hire_date = employee.find('hire_date').text
    # Insert data into the employees_XML table
    mycursor.execute("""INSERT INTO employees_XML
    (employee_name, department, salary, hire_date)
    VALUES (%s, %s, %s, %s)""", (employee_name, department,
    salary, hire_date))

# Commit the changes after all rows are inserted
mydb.commit()
# Close the cursor and connection
mycursor.close()
mydb.close()
```
