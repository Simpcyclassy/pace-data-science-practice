**Five NumPy Functions:**

2. `np.concatenate()`

**Three SciPy Functions:**

## **NumPy Functions:**


### `np.reshape()`:
This function reshapes an existing array into a new shape without changing its data.

**Observation**: This is helpful when you need to change the shape of data, such as for matrix operations or image data.


```python
import numpy as np

# Original array
array = np.array([1, 2, 3, 4, 5, 6])
reshaped_array = np.reshape(array, (2, 3))
print("Original Array:", array)
print("Reshaped Array (2x3):\n", reshaped_array)
```

### `np.concatenate()`:
This function joins two or more arrays along a specified axis.

**Observation**: Useful when combining datasets or merging information.

```python
import numpy as np

array1 = np.array([[1, 2], [3, 4]])
array2 = np.array([[5, 6]])

# Concatenating arrays along axis 0
result = np.concatenate((array1, array2), axis=0)
print("Concatenated Array:\n", result)
```

### `np.random.randint()`:
Generates random integers within a specified range.

**Observation**: Helpful for creating random samples or initializing weights in machine learning.

```python
import numpy as np

random_array = np.random.randint(1, 10, size=(3, 3))
print("Random 3x3 Array:\n", random_array)
```

### `np.mean()`:
Calculates the mean (average) of array elements.

**Observation**: Commonly used in data analysis for finding the central tendency.

```python
import numpy as np

array = np.array([10, 20, 30, 40, 50])
mean_value = np.mean(array)
print("Mean Value:", mean_value)
```

### `np.linspace()`:
Generates evenly spaced numbers over a specified interval.

**Observation**: Great for creating sequences, especially for plotting purposes.

```python
import numpy as np

linspace_array = np.linspace(0, 10, num=5)
print("Linspace Array:", linspace_array)
```

## **SciPy Functions:**


### `scipy.optimize.minimize()`:
Minimizes a given function.

**Observation**: Useful for optimization problems in machine learning and data fitting.

```python
from scipy.optimize import minimize

# Function to minimize: (x - 3)^2
def func(x):
    return (x - 3)**2

result = minimize(func, x0=0)
print("Minimum value found at:", result.x)
```

### `scipy.integrate.quad()`:
Performs definite integration.

**Observation**: Integration is often used in physics simulations and other areas involving calculus.

```python
from scipy.integrate import quad

# Function to integrate: x^2
def integrand(x):
    return x**2

result, error = quad(integrand, 0, 3)
print("Result of integration from 0 to 3:", result)
```

### `scipy.stats.norm()`:
Generates a normal (Gaussian) distribution.

**Observation**: Very useful in statistics for modeling and analyzing data that follows a normal distribution.

```python
from scipy.stats import norm
import matplotlib.pyplot as plt

# Generate random numbers from a normal distribution
data = norm.rvs(size=1000, loc=0, scale=1)

# Plot the histogram
plt.hist(data, bins=30, density=True)
plt.title("Histogram of Normal Distribution")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()
```
