```python
import pandas as pd
from sodapy import Socrata

# Unauthenticated client only works with public data sets. 
# Note no username or password:
client = Socrata("data.winnipeg.ca", "dOrQbTCqCNLTutDyoQLwoGjNE")


# Getting data
wpg_inspect_defects = client.get("ytpf-7jye")

# Convert to pandas DataFrame
wpg_inspect_defects_df = pd.DataFrame.from_records(wpg_inspect_defects)
```

```python
wpg_inspect_defects_df['inspection_count'] = pd.to_numeric(wpg_inspect_defects_df['inspection_count'],errors = 'coerce')
wpg_inspect_defects_df['permit_count'] = pd.to_numeric(wpg_inspect_defects_df['permit_count'],errors = 'coerce')
wpg_inspect_defects_df['defect_count'] = pd.to_numeric(wpg_inspect_defects_df['defect_count'],errors = 'coerce')
wpg_inspect_defects_df['year'] = pd.to_numeric(wpg_inspect_defects_df['year'],errors = 'coerce')
```

```python
#wpg_inspect_defects_df_drop = wpg_inspect_defects_df.drop('location', axis=1, inplace=False)

wpg_inspect_defects_df.drop('location', axis=1, inplace=True)
wpg_inspect_defects_df.head()
```

```python
wpg_inspect_defects_df.shape
```

```python
wpg_inspect_defects_df_pv = pd.pivot_table(wpg_inspect_defects_df, values=['inspection_count','defect_count'],
                        index=['ward','year'], columns=['inspection_type'])
wpg_inspect_defects_df_pv
```

```python
wpg_inspect_defects_df_pv.shape
```

```python
wpg_inspect_defects_df_pv.columns.map('_'.join)
```

```python
wpg_inspect_defects_df_pv.columns = wpg_inspect_defects_df_pv.columns.map('_'.join)
wpg_inspect_defects_df_pv
```

```python
wpg_inspect_defects_df_pv = wpg_inspect_defects_df_pv.reset_index()
wpg_inspect_defects_df_pv
```

```python
wpg_inspect_defects_df_pv['total_inspection_count'] = wpg_inspect_defects_df_pv['inspection_count_Residential Building'] + wpg_inspect_defects_df_pv['inspection_count_Residential Electrical'] + wpg_inspect_defects_df_pv['inspection_count_Residential Plumbing']
wpg_inspect_defects_df_pv['total_defect_count'] = wpg_inspect_defects_df_pv['defect_count_Residential Building'] + wpg_inspect_defects_df_pv['defect_count_Residential Electrical'] + wpg_inspect_defects_df_pv['defect_count_Residential Plumbing']
wpg_inspect_defects_df_pv['total_defect_ratio'] = wpg_inspect_defects_df_pv['total_defect_count']/wpg_inspect_defects_df_pv['total_inspection_count']
wpg_inspect_defects_df_pv
```

```python
wpg_inspect_defects_target = wpg_inspect_defects_df_pv[['ward','year','total_defect_ratio']]
wpg_inspect_defects_target
```

```python
wpg_inspect_defects_target.info()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(16, 8))
sns.set_theme(style="whitegrid")

# Plot the responses for different events and regions
sns.lineplot(x="year", y="total_defect_ratio",
             hue="ward", 
             data=wpg_inspect_defects_target, 
             palette=sns.color_palette("deep", 20),
            linewidth=2.5)
plt.show()
```
